##month buckets

```python
df["month"]="January"
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
df.loc[df["transaction_date"] > pd.Timestamp("2023-01-31"), "month"] = "February"
df.loc[df["transaction_date"] > pd.Timestamp("2023-02-28"), "month"] = "March"
df.loc[df["transaction_date"] > pd.Timestamp("2023-03-31"), "month"] = "April"
df.loc[df["transaction_date"] > pd.Timestamp("2023-04-30"), "month"] = "May"
df.loc[df["transaction_date"] > pd.Timestamp("2023-05-31"), "month"] = "June"
df
```

##converting transaction_time buckets

```python
import datetime

df["time_bucket"] = "morning"

# Convert 'transaction_time' to datetime objects. It is currently of object Dtype (strings like 'HH:MM:SS').
df["transaction_time_dt"] = pd.to_datetime(df["transaction_time"], format="%H:%M:%S")

# Extract only the time component for comparison
time_only = df["transaction_time_dt"].dt.time

# Define time boundaries using datetime.time objects for accurate comparison
afternoon_start = datetime.time(12, 0, 0) # 12:00:00
evening_start = datetime.time(17, 0, 0)   # 17:00:00
night_start = datetime.time(19, 0, 0)     # 19:00:00

df.loc[time_only >= afternoon_start, "time_bucket"] = "afternoon"
df.loc[time_only >= evening_start, "time_bucket"] = "evening"
df.loc[time_only >= night_start, "time_bucket"] = "night"
df
```

##price bucket

```python
df["price_bucket"]="low"
df.loc[df["unit_price"]>15,"price_bucket"]="average"
df.loc[df["unit_price"]>30,"price_bucket"]="high"
df
```

##quantity_bucket

```python
df["quantity_bucket"]="retail"
df.loc[df["transaction_qty"]>3,"quantity_bucket"]="medium"
df.loc[df["transaction_qty"]>5,"quantity_bucket"]="wholesale"
df
```
